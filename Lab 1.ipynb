### *Activity 1*

```python
import string
def cleaned_string(paragraph: str) -> list[str]:
    words = paragraph.split()
    cleaned = [word.strip(string.punctuation).lower() for word in words]
    return [word for word in cleaned if word]

text = "The morning fog made everything feel mysterious. A young artist walked slowly, thinking about the promise. It was time to move forward."
print(cleaned_string(text))
```

### *Activity 2*

```python
import string
from collections import Counter

def word_count(text:str):
    words = [word.strip(string.punctuation).lower() 
        for word in text.split()]
    words = [w for w in words if w]

    counts = Counter(words)
    common_words =counts.most_common(10)

    print(f"{'Word':<25} | {'Count':<5}")
    print("-" * 28)
    a =0
    for word, count in common_words:
        a = a+1
        print(a,f"{word:<23} | {count:<5}")
        

sample = """
Data science is an interdisciplinary field that uses scientific methods, processes, 
algorithms and systems to extract knowledge and insights from noisy, structured and 
unstructured data, and apply knowledge from data across a broad range of application domains. 
Data science is related to data mining, machine learning and big data.
"""

word_count(sample)
```

### *Activity 3*

```python
import re
import pandas as pd
paragraph = ("Natural language processing is a fascinating field. "
    "It bridges the gap between human communication and computer understanding. "
    "Python makes text processing very simple and efficient")

sentences = [s.strip() for s in re.split(r'(?<=[.!?])\s+', paragraph) if s.strip()]

df = pd.DataFrame({'sentence': sentences})
df['word_count'] = df['sentence'].apply(lambda s: len(s.split()))
df['char_count'] = df['sentence'].apply(len)
print(df)
```

### *Activity 4*

```python
import numpy as np

def cosine_similarity(v1: np.ndarray, v2: np.ndarray) -> float:
    dot_product = np.dot(v1, v2)
    norm_v1 = np.linalg.norm(v1)
    norm_v2 = np.linalg.norm(v2)
    
    if norm_v1 == 0 or norm_v2 == 0:
        return 0.0
    
    return dot_product / (norm_v1 * norm_v2)

pair_A = (np.array([3, 2, 0, 1]), np.array([3, 1, 1, 1]))
pair_B = (np.array([1, 0, 0, 5]), np.array([0, 4, 2, 0]))

sim_A = cosine_similarity(pair_A[0], pair_A[1])
sim_B = cosine_similarity(pair_B[0], pair_B[1])

print(f"Cosine Similarity Pair A: {sim_A:.4f}")
print(f"Cosine Similarity Pair B: {sim_B:.4f}")
```

### *Activity 5*

```python
import json
import string
from collections import Counter
import pandas as pd

# 1. Load raw JSON
with open('reviews.json', 'r') as f:
    data = json.load(f)

# 2. Build DataFrame
df = pd.DataFrame(data)

# function to find the most common word in a single string
def common_word(text: str) -> str:
    cleaned = [word.strip(string.punctuation).lower() for word in text.split()]
    cleaned = [w for w in cleaned if w]
    if not cleaned:
        return ""
    return Counter(cleaned).most_common(1)[0][0]

# 3. Add most common word and word count columns
df['common_word'] = df['review'].apply(common_word)
df['word_count'] = df['review'].apply(lambda x: len(x.split()))

# 4. Sort DataFrame by word count in descending order
df_sorted = df.sort_values(by='word_count', ascending=False).reset_index(drop=True)
pd.set_option('display.width', 1000) #format
pd.set_option('display.max_columns', None) #format
print(df_sorted)
```
