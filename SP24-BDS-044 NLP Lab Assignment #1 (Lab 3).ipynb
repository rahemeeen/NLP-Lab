## **Task 1:** Regex fundamentals (character classes & quantifiers)

```python
text = "In 2024, Alice visited Paris on 15th March. " \
"She bought 3 gifts for $120 and 2 books." \
"The year 1998 was special! Emails: alice@mail.com, bob123@yahoo.com. Call 0301-2345678."
print(text)
```

#### a) Extract all digits from the text

```python
digits = re.findall(r'\d', text)
print(digits)
```

#### b) Extract all words that start with a capital letter

```python
capital_words = re.findall(r'\b[A-Z]\w*', text)
print(capital_words)
```

#### c) Extract all sequences of exactly 4 digits (e.g., years)

```python
years = re.findall(r'\b\d{4}\b', text)
print(years)
```

#### d) Extract all words that are between 3 to 6 characters long

```python
words = re.findall(r'\b\w{3,6}\b', text)
print(words)
```

## **Task 2:** Pattern-based extraction from real-world data

```python
paragraph = """Dear customer, your order ORD-4521 was placed on 12/03/2025 with a 10% discount.
Contact support at help.desk@shopline.com or call 0345-6789012 for urgent issues.
Another customer, sara_99@gmail.com, was charged Rs. 1500 on 01/09/2024 and received a 12.5% refund.
Please reply to billing@shopline.com before 31/12/2025. 
A 5% service fee applies.Backup contact: 0333-1122334."""
print(paragraph)
```

#### a) Write a regex pattern using re.findall() to extract all email addresses

```python
emails = re.findall(r'[\w.+-]+@[\w-]+\.[\w.]+', paragraph)
print(emails)
```

#### b) Write a regex pattern to extract all phone numbers (assume format: 03XX-XXXXXXX)

```python
phones = re.findall(r'03\d{2}-\d{7}', paragraph)
print(phones)
```

#### c) Write a regex pattern to extract all dates in DD/MM/YYYY format

```python
dates = re.findall(r'\b\d{2}/\d{2}/\d{4}\b', paragraph)
print(dates)
```

#### d) Write a regex pattern to extract all percentage values (e.g., 45%, 12.5%)

```python
percentages = re.findall(r'\d+(?:\.\d+)?%', paragraph)
print(percentages)
```

## **Task 3:** Word tokenization — NLTK vs spaCy

```python
sample_para = "I don't think it's a good idea, but I'll try. " \
"It's raining heavily, isn't it? The company's revenue grew by 20% in 2024. " \
"Well-done, John! We can't wait for the results; they'll be announced tomorrow."
print(sample_para)
```

#### a) Tokenize with NLTK's `word_tokenize()`

```python
import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk_tokens = nltk.word_tokenize(sample_para)
print(nltk_tokens)
```

#### b) Tokenize the same paragraph using a blank spaCy pipeline

```python
import spacy
blank_nlp = spacy.blank("en")
doc_blank = blank_nlp(sample_para)
spacy_tokens = [token.text for token in doc_blank]
print(spacy_tokens)
```

#### c) Compare and write down (in a comment) at least 2 differences you noticed between the two outputs

```python
# Difference 1: Hyphenated Words
# NLTK treats hyphenated words as a single token (e.g., 'Well-done' -> ['Well-done']).
# spaCy splits hyphenated words into separate tokens and punctuation (e.g., 'Well-done' -> ['Well', '-', 'done']).

# Difference 2: Punctuation & Contraction Handling
# While both frameworks split common contractions (e.g., "don't" -> ["do", "n't"]), 
# spaCy isolates punctuation marks strictly as independent tokens across all edge cases, 
# resulting in a higher overall token count for complex sentences.
```

#### d) Count the total number of tokens produced by each method

```python
print(f"NLTK total tokens : {len(nltk_tokens)}")
print(f"spaCy total tokens: {len(spacy_tokens)}")
```
